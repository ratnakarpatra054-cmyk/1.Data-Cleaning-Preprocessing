# Import the Dataset Using Pandas

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("Dataset.csv")
```

```python
df.head()
```

# Identify missing and duplicate records

```python
# Check dataset rows and columns
print("Dataset Shape:", df.shape)

# Check missing values
print(df.isnull().sum())

# check duplicate records
print(df.duplicated().sum())
```

# Handle null values and inconsistent formats

```python
# create a copy of original dataset
clean_df = df.copy()

# Remove duplicate records
clean_df = clean_df.drop_duplicates()

# Remove extra spaces from text columns
text_columns = clean_df.select_dtypes(include="object").columns

for col in text_columns: clean_df[col] = clean_df[col].str.strip()
```

```python
# handle missing values
clean_df["director"] = clean_df["director"].fillna("Unknown")
clean_df["title"] = clean_df["title"].fillna("Unknown")
clean_df["country"] = clean_df["country"].fillna("Unknown")
clean_df["rating"] = clean_df["rating"].fillna("Unknown")

# Check missing values after cleaning
print(clean_df.isnull().sum())
```

# Transform categorical and date-related columns

```python
# convert data_added into datetime
clean_df["date_added"] = pd.to_datetime(
    clean_df["date_added"],
    errors="coerce"
)
# Extract year and month
clean_df["date_added_year"] = (
    clean_df["date_added"].dt.year
)
clean_df["date_added_month"] = (
    clean_df["date_added"].dt.month
)

print(clean_df[[
    "title",
    "date_added",
    "date_added_year",
    "date_added_month"
]].head())
```

```python
#  Separate duration value and unit

duration_split = clean_df["duration"].str.extract(
    r"(\d+)\s*(.*)"
)

clean_df["duration_value"] = pd.to_numeric(
    duration_split[0],
    errors="coerce"
)

clean_df["duration_unit"] = duration_split[1]

print(clean_df[[
    "title",
    "duration",
    "duration_value",
    "duration_unit"
]].head(10))
```

```python
# Standardize categorical columns

clean_df["type"] = clean_df["type"].str.title()

clean_df["rating"] = clean_df["rating"].str.upper()

# Convert categorical columns to category datatype
clean_df["type"] = clean_df["type"].astype("category")

clean_df["rating"] = clean_df["rating"].astype("category")

print(clean_df[["type", "rating"]].head())

print(clean_df.dtypes)
```

# Create and save the clean dataset

```python
# Final dataset verification

print("Original Dataset Shape:", df.shape)

print("Cleaned Dataset Shape:", clean_df.shape)

print(clean_df.isnull().sum())

print(clean_df.duplicated().sum())
```

```python
# Export cleaned dataset

clean_df.to_csv(
    "netflix_cleaned.csv",
    index=False
)

print("File saved as netflix_cleaned.csv")
```
